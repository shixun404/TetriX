```python
import networkx as nx
import json
import pickle as pkl

# Load the node mapping (IP:PORT → idx)
# with open('100_best_graph.pkl', 'rb') as f:
with open('../FABRIC_400_best_test_graph.pkl', 'rb') as f:
    graph = pkl.load(f)
# with open("../node_mapping_100.pkl", "rb") as f:
with open("../node_mapping_400.pkl", "rb") as f:
    node_mapping = pkl.load(f)

# Invert the mapping to get (idx → IP:PORT)
idx_to_ip = {idx: ip_port for ip_port, idx in node_mapping.items()}


# Prepare the outgoing connections mapping
outgoing_connections = {}

for node in graph.nodes():
    if node in idx_to_ip:
        node_ip = idx_to_ip[node]  # Convert idx back to IP:PORT
        neighbors = list(graph.successors(node))  # Outgoing neighbors
        # Convert neighbor indices back to IP:PORT
        neighbor_ips = [idx_to_ip[neighbor] for neighbor in neighbors if neighbor in idx_to_ip]
        # Store the outgoing connections
        outgoing_connections[node_ip] = neighbor_ips

# Save as JSON file
with open("outgoing_connections.json", "w") as f:
    json.dump(outgoing_connections, f, indent=4)
```
